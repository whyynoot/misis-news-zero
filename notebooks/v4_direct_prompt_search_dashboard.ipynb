# V4 Direct Prompt Search Dashboard

Reads asynchronous experiment outputs and ranks models/prompts by direct-only and broader strength metrics.

```
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
OUT = ROOT / 'analysis_outputs'
metrics_path = OUT / 'v4_strength_metrics.csv'
manifest_path = OUT / 'v4_strength_manifest.csv'
print(metrics_path)
print(manifest_path)
```

```
if not metrics_path.exists():
    raise FileNotFoundError('Run scripts/evaluate_v4_strength_metrics.py after at least one prediction file exists')
metrics = pd.read_csv(metrics_path)
manifest = pd.read_csv(manifest_path) if manifest_path.exists() else pd.DataFrame()
print('runs', metrics['run_id'].nunique(), 'rows', len(metrics))
metrics.head()
```

## Top Direct

```
cols = ['model','prompt','n_news','threshold','micro_precision','micro_recall','micro_f1','macro_f1_supported','sample_f1','mean_pred_labels','file_name']
(metrics[metrics.scope.eq('direct')]
 .sort_values(['micro_f1','micro_recall','n_news'], ascending=False)
 [cols]
 .head(30))
```

## Direct Recall First

```
(metrics[metrics.scope.eq('direct')]
 .sort_values(['micro_recall','micro_f1','mean_pred_labels'], ascending=[False,False,True])
 [cols]
 .head(30))
```

## Direct + Context

```
(metrics[metrics.scope.eq('direct_context')]
 .sort_values(['micro_f1','micro_recall','n_news'], ascending=False)
 [cols]
 .head(30))
```

## Context / Weak Diagnostics

```
for scope in ['context','weak','all']:
    display(scope)
    display((metrics[metrics.scope.eq(scope)]
      .sort_values(['micro_f1','micro_recall','n_news'], ascending=False)
      [cols]
      .head(15)))
```
